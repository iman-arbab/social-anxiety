برای نمایش نمودار تک متغیره کتگوریکال نمودار های میله ای, pie و treemap امکان پذیر و تا حدی منطقی به نظر می رسد.

```python
def categorical_univariate_all_plots(df, column):

    counts = df[column].value_counts().reset_index()

    fig = make_subplots(rows=1, cols=3, subplot_titles=["Bar Chart", "Pie Chart", "Treemap"], specs=[[{"type": "xy"}, {"type": "domain"}, {"type": "domain"}]])

    bar = px.bar(counts, x=column, y="count")

    for trace in bar.data:
        fig.add_trace(trace, row=1, col=1)

    pie = px.pie(counts, names=column, values="count")

    for trace in pie.data:
        fig.add_trace(trace, row=1, col=2)

    treemap = px.treemap(counts, path=[column], values="count")

    for trace in treemap.data:
        fig.add_trace(trace, row=1, col=3)

    fig.update_layout(
        title_text=f"Univariate Categorical Analysis: {column}",
        height=500
    )

    return fig
```

```python
def categorical_univariate_plot(df, column, plot_type="bar"):
    counts = df[column].value_counts().reset_index()

    if plot_type == "bar":
        fig = px.bar(counts, x=column, y="count", title=f"Distribution of {column}", labels={"count": "Count"})


    elif plot_type == "pie":
        fig = px.pie(counts, names=column, values="count", title=f"Distribution of {column}")


    elif plot_type == "treemap":
        fig = px.treemap(counts, path=[column], values="count", title=f"Distribution of {column}")

    return fig

```

تست نمودار های تک متغیره با داده های تولید شده ی رندم:

```python
fig = categorical_univariate_all_plots(df, "Gender")
fig.show()
```

```python
fig1 = categorical_univariate_plot(df, "Gender", "bar")
fig1.show()


fig2 = categorical_univariate_plot(df, "City", "pie")
fig2.show()


fig3 = categorical_univariate_plot(df, "Payment_Type", "treemap")
fig3.show()
```

برای نمایش نمودار دو متغیره کتگوریکال نمودار های میله ای گروهی, stacked bar و heatmap امکان پذیر و تا حدی منطقی به نظر می رسد.

```python
def categorical_bivariate_all_plots(df, column1, column2):

    table = pd.crosstab(df[column1], df[column2])

    fig = make_subplots(rows=1, cols=3,subplot_titles=["Grouped Bar", "Stacked Bar", "Heatmap"], specs=[[{"type": "xy"}, {"type": "xy"}, {"type": "heatmap"}]])

    grouped = px.histogram(df, x=column1, color=column2, barmode="group")

    for trace in grouped.data:
        fig.add_trace(trace, row=1, col=1)

    stacked = px.histogram(df, x=column1, color=column2, barmode="stack")

    for trace in stacked.data:
        fig.add_trace(trace, row=1, col=2)

    heatmap = px.imshow(table, text_auto=True)

    for trace in heatmap.data:
        fig.add_trace(trace, row=1, col=3)

    fig.update_layout(title_text = f"Bivariate Categorical Analysis: {column1} vs {column2}", height=500)

    return fig
```

```python
def categorical_bivariate_plot(df, column1, column2, plot_type="grouped_bar"):
    if plot_type == "grouped_bar":
        fig = px.histogram(df, x=column1, color=column2, barmode="group", title=f"{column1} vs {column2}")


    elif plot_type == "stacked_bar":
        fig = px.histogram(df, x=column1, color=column2, barmode="stack", title=f"{column1} vs {column2}")

    elif plot_type == "heatmap":
        table = pd.crosstab(df[column1], df[column2])
        fig = px.imshow(table, text_auto=True, title=f"{column1} vs {column2}")

    return fig
```

تست نمودار های تک متغیره با داده های تولید شده ی رندم:

```python
fig = categorical_bivariate_all_plots(df, "Gender", "Payment_Type")
fig.show()
```

```python
fig4 = categorical_bivariate_plot(
    df,
    "Gender",
    "Payment_Type",
    "grouped_bar"
)
fig4.show()


fig5 = categorical_bivariate_plot(
    df,
    "Gender",
    "Payment_Type",
    "stacked_bar"
)
fig5.show()


fig6 = categorical_bivariate_plot(
    df,
    "City",
    "Payment_Type",
    "heatmap"
)
fig6.show()

```

برای نمایش نمودار دو متغیره کتگوریکال نمودار های facetted_bar و sunburst و treemap و parallel_categories امکان پذیر و تا حدی منطقی به نظر می رسد.

```python
def categorical_trivariate_all_plots(df, column1, column2, column3):

    fig = make_subplots(
        rows=2,
        cols=2,
        subplot_titles=[
            "Faceted Bar",
            "Sunburst",
            "Treemap",
            "Parallel Categories"
        ],
        specs=[
            [{"type": "xy"}, {"type": "domain"}],
            [{"type": "domain"}, {"type": "domain"}]
        ]
    )

    faceted = px.histogram(df, x=column1, color=column2, facet_col=column3, barmode="group")

    for trace in faceted.data:
        fig.add_trace(trace, row=1, col=1)

    sunburst = px.sunburst(df, path=[column1, column2, column3])

    for trace in sunburst.data:
        fig.add_trace(trace, row=1, col=2)

    treemap = px.treemap(df, path=[column1, column2, column3])

    for trace in treemap.data:
        fig.add_trace(trace, row=2, col=1)

    parallel = px.parallel_categories(df, dimensions=[column1, column2, column3])

    for trace in parallel.data:
        fig.add_trace(trace, row=2, col=2)

    fig.update_layout(
        title_text=f"Trivariate Categorical Analysis: {column1}, {column2} and {column3}",
        height=900
    )

    return fig
```

```python
def categorical_trivariate_plot(df, column1, column2, column3, plot_type="facetted_bar"):
    if plot_type == "facetted_bar":
        fig = px.histogram(df, x=column1, color=column2, facet_col=column3, barmode="group", title=f"{column1}, {column2} and {column3}")

    elif plot_type == "sunburst":
        fig = px.sunburst(df, path=[column1, column2, column3], title=f"{column1} to {column2} to {column3}")

    elif plot_type == "treemap":
        fig = px.treemap(df, path=[column1, column2, column3], title=f"{column1} to {column2} to {column3}")

    elif plot_type == "parallel_categories":

        fig = px.parallel_categories(df, dimensions=[column1, column2, column3], title=f"{column1}, {column2} and {column3}")

    return fig
```

تست نمودار های سه متغیره با داده های تولید شده ی رندم:

```python
fig = categorical_trivariate_all_plots(df, "Gender", "City", "Payment_Type")
fig.show()
```

```python

fig7 = categorical_trivariate_plot(
    df,
    "Gender",
    "Payment_Type",
    "Class",
    "facetted_bar"
)

fig7.show()


fig8 = categorical_trivariate_plot(
    df,
    "City",
    "Payment_Type",
    "Class",
    "sunburst"
)

fig8.show()


fig9 = categorical_trivariate_plot(
    df,
    "City",
    "Payment_Type",
    "Class",
    "treemap"
)

fig9.show()


fig10 = categorical_trivariate_plot(
    df,
    "Gender",
    "City",
    "Payment_Type",
    "parallel_categories"
)

fig10.show()
```
