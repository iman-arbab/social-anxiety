```python
# a function for 1 column

def categorical_relationship_one_column(df, col1, col2, alpha=0.05):
    print(f"Analyzing relationship between [{col1}] and [{col2}]\n")
    
    print("Statistical hypotheses:")
    print(f"H0: Variables '{col1}' and '{col2}' are completely independent (no relationship).")
    print(f"H1: Variables '{col1}' and '{col2}' are dependent.\n")
    
    contingency_table = pd.crosstab(df[col1], df[col2])
    
    chi2, p_value, dof, expected = stats.chi2_contingency(contingency_table)
    
    n = contingency_table.to_numpy().sum()
    min_dim = min(contingency_table.shape[0] - 1, contingency_table.shape[1] - 1)
    cramers_v = np.sqrt(chi2 / (n * min_dim))
    
    print("Test results:")
    print(f"Chi-square value: {chi2:.4f}")
    print(f"P-value: {p_value:.8f}")
    print(f"Degrees of freedom (DoF): {dof}")
    print(f"Cramer's V (strength of association): {cramers_v:.4f} (0 - 1)\n")
    
    print("Conclusion:")
    if p_value < alpha:
        print(f"H0 is rejected (P-value < {alpha}).")
        print(f"Result: There is sufficient statistical evidence to show a significant relationship between '{col1}' and '{col2}'.")
    else:
        print(f"H0 is not rejected (P-value >= {alpha}).")
        print(f"Result: '{col1}' and '{col2}' appear to be independent.")
```

```python
categorical_relationship_one_column(df, 'Inherited Risk Triggered', 'Target')
```

```python
# a function for 2 columns

def categorical_relationship_two_columns(df, col1, val1, col2, val2, target_col, alpha=0.05):
    segments = f"'{val1} {col1}' & '{val2} {col2}'"
    print(f"Analyzing [{segments}] vs [{target_col}]")
    
    condition = (df[col1] == val1) & (df[col2] == val2)
    
    temp_groups = np.where(condition, "Target Group", "Others")
    
    contingency_table = pd.crosstab(temp_groups, df[target_col])
    print("Contingency Table:")
    print(contingency_table, "\n")

    chi2, p_value, dof, expected = stats.chi2_contingency(contingency_table)
    
    n = contingency_table.to_numpy().sum()
    min_dim = min(contingency_table.shape[0] - 1, contingency_table.shape[1] - 1)
    cramers_v = np.sqrt(chi2 / (n * min_dim))
    
    print("Test results:")
    print(f"Chi-square value: {chi2:.4f}")
    print(f"P-value: {p_value:.8f}")
    print(f"Degrees of freedom (DoF): {dof}")
    print(f"Cramer's V: {cramers_v:.4f} (0 - 1)\n")
    
    print("Conclusion:")
    if p_value < alpha:
        print(f"H0 is rejected (P-value < {alpha}).")
        print(f"Result: Significant relationship between [{segments}] and ['{target_col}'].")
    else:
        print(f"H0 is not rejected (P-value >= {alpha}).")
        print(f"Result: No significant relationship between [{segments}] and ['{target_col}'].")
```

```python
categorical_relationship_two_columns(df, 'Family History of Anxiety', 'Yes', 'Family History of Anxiety', 'Yes', 'Target')
```

```python
def anova(df, c_col, n_col, alpha=0.05):
    print(f"Analyzing Variance (ANOVA) between [{c_col}] and [{n_col}]\n")
    
    print("Statistical hypotheses:")
    print(f"H0: The mean of '{n_col}' is the same across all groups of '{c_col}'.")
    print(f"H1: The mean of '{n_col}' is different for at least one group of '{c_col}'.\n")
    
    temp_df = df[[c_col, n_col]]

    groups = []

    for name, group in temp_df.groupby(c_col):

        group_values = group[n_col].values
        groups.append(group_values)

    f_stat, p_value = stats.f_oneway(*groups)

    print("Test results:")
    print(f"F-statistic: {f_stat:.4f}")
    print(f"P-value: {p_value:.8f}\n")

    print("Conclusion:")
    if p_value < alpha:
        print(f"H0 is rejected (P-value < {alpha}).")
        print(f"Result: There is a statistically significant difference in the mean of '{n_col}' across different '{c_col}' groups.")
    else:
        print(f"H0 is not rejected (P-value >= {alpha}).")
        print(f"Result: There is NO significant difference in the mean of '{n_col}' across the groups.")
```

```python
anova(df, 'Gender', 'Breathing Rate (breaths/min)')
```
