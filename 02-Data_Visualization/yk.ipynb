```python
def categorical_univariate_all_plots(df, column):

    counts = df[column].value_counts().reset_index()

    fig = make_subplots(rows=1, cols=3, subplot_titles=["Bar Chart", "Pie Chart", "Treemap"], specs=[[{"type": "xy"}, {"type": "domain"}, {"type": "domain"}]])

    bar = px.bar(counts, x=column, y="count", color=column, color_discrete_sequence=COLORS, text="count")

    for trace in bar.data:
        fig.add_trace(trace, row=1, col=1)


    pie = px.pie(counts, names=column, values="count", color=column, color_discrete_sequence=COLORS, hole=0.45)

    for trace in pie.data:
        fig.add_trace(trace, row=1, col=2)

    treemap = px.treemap(counts, path=[column], values="count")

    for trace in treemap.data:
        fig.add_trace(trace, row=1, col=3)

    fig.update_layout(
        title_text=f"Univariate Categorical Analysis: {column}",
        height=500
    )

    return fig
```

```python
def categorical_univariate_plot(df, column, plot_type="bar"):
    counts = df[column].value_counts().reset_index()

    if plot_type == "bar":
        fig = px.bar(counts, x=column, y="count", title=f"Distribution of {column}", labels={"count": "Count"})


    elif plot_type == "pie":
        fig = px.pie(counts, names=column, values="count", title=f"Distribution of {column}")


    elif plot_type == "treemap":
        fig = px.treemap(counts, path=[column], values="count", title=f"Distribution of {column}")

    return fig

```

```python
def categorical_bivariate_all_plots(df, column1, column2, x_order = ["Low", "Medium", "High"]):

    table = pd.crosstab(df[column1], df[column2])

    fig = make_subplots(rows=1, cols=3,subplot_titles=["Grouped Bar", "Stacked Bar", "Heatmap"], specs=[[{"type": "xy"}, {"type": "xy"}, {"type": "heatmap"}]])

    if x_order is None:
        x_order = df[column1].dropna().unique().tolist()
    x_order = [value for value in x_order if value in df[column1].unique()]

    grouped = px.histogram(df, x=column1, color=column2, category_orders={column1: x_order})

    for i, trace in enumerate(grouped.data):
        trace.offsetgroup = f"group_{i}"
        fig.add_trace(trace, row=1, col=1)

    fig.update_xaxes(title_text=column1, categoryorder="array", categoryarray=x_order, row=1, col=1)
    fig.update_yaxes(title_text="Count", row=1, col=1)

    stacked = px.histogram(df, x=column1, color=column2, barmode="stack")

    for trace in stacked.data:
        trace.offsetgroup = "stacked"
        fig.add_trace(trace, row=1, col=2)
    fig.update_xaxes(title_text=column1, row=1, col=2)
    fig.update_yaxes(title_text=column2, row=1, col=2)

    heatmap = px.imshow(table, text_auto=True, color_continuous_scale="Blues", labels={"x": column2, "y": column1, "color": "Count"})

    for trace in heatmap.data:
        fig.add_trace(trace, row=1, col=3)

    fig.update_xaxes(title_text=column2, row=1, col=3)
    fig.update_yaxes(title_text=column1, categoryorder="array", categoryarray=x_order, row=1, col=3)

    fig.update_layout(title_text = f"Bivariate Categorical Analysis: {column1} vs {column2}", height=500)

    return fig
```

```python
def categorical_bivariate_plot(df, column1, column2, plot_type="grouped_bar"):
    if plot_type == "grouped_bar":
        fig = px.histogram(df, x=column1, color=column2, barmode="group", title=f"{column1} vs {column2}")


    elif plot_type == "stacked_bar":
        fig = px.histogram(df, x=column1, color=column2, barmode="stack", title=f"{column1} vs {column2}")

    elif plot_type == "heatmap":
        table = pd.crosstab(df[column1], df[column2])
        fig = px.imshow(table, text_auto=True, title=f"{column1} vs {column2}")

    return fig
```

برای نمایش نمودار دو متغیره کتگوریکال نمودار های facetted_bar و sunburst و treemap و parallel_categories امکان پذیر و تا حدی منطقی به نظر می رسد.

```python
def categorical_trivariate_all_plots(df, column1, column2, column3):

    fig = make_subplots(
        rows=2,
        cols=2,
        subplot_titles=[
            "Faceted Bar",
            "Sunburst",
            "Treemap",
            "Parallel Categories"
        ],
        specs=[
            [{"type": "xy"}, {"type": "domain"}],
            [{"type": "domain"}, {"type": "domain"}]
        ]
    )

    faceted = px.histogram(df, x=column1, color=column2, facet_col=column3, barmode="group")

    for trace in faceted.data:
        fig.add_trace(trace, row=1, col=1)

    sunburst = px.sunburst(df, path=[column1, column2, column3])

    for trace in sunburst.data:
        fig.add_trace(trace, row=1, col=2)

    treemap = px.treemap(df, path=[column1, column2, column3])

    for trace in treemap.data:
        fig.add_trace(trace, row=2, col=1)

    parallel = px.parallel_categories(df, dimensions=[column1, column2, column3])

    for trace in parallel.data:
        fig.add_trace(trace, row=2, col=2)

    fig.update_layout(
        title_text=f"Trivariate Categorical Analysis: {column1}, {column2} and {column3}",
        height=900
    )

    return fig
```

```python
def categorical_trivariate_plot(df, column1, column2, column3, plot_type="facetted_bar"):
    if plot_type == "facetted_bar":
        fig = px.histogram(df, x=column1, color=column2, facet_col=column3, barmode="group", title=f"{column1}, {column2} and {column3}")

    elif plot_type == "sunburst":
        fig = px.sunburst(df, path=[column1, column2, column3], title=f"{column1} to {column2} to {column3}")

    elif plot_type == "treemap":
        fig = px.treemap(df, path=[column1, column2, column3], title=f"{column1} to {column2} to {column3}")

    elif plot_type == "parallel_categories":

        fig = px.parallel_categories(df, dimensions=[column1, column2, column3], title=f"{column1}, {column2} and {column3}")

    return fig
```

```python

np.random.seed(42)

n = 300

df_numeric = pd.DataFrame({
    "Age": np.random.normal(
        loc=30,
        scale=8,
        size=n
    ),

    "Height": np.random.normal(
        loc=175,
        scale=7,
        size=n
    ),

    "Weight": np.random.normal(
        loc=70,
        scale=10,
        size=n
    ),

    "Income": np.random.normal(
        loc=5000,
        scale=1500,
        size=n
    )
})
```

```python
def numerical_univariate_all_plots(df, column):

    fig = make_subplots(
        rows=1,
        cols=5,
        subplot_titles=["Histogram", "Box Plot", "Violin Plot", "ECDF", "Strip Plot"],
        specs=[[{"type": "xy"}, {"type": "xy"}, {"type": "xy"}, {"type": "xy"}, {"type": "xy"}]]
    )

    histogram = px.histogram(df, x=column)

    for trace in histogram.data:
        fig.add_trace(trace, row=1, col=1)


    box = px.box(df, y=column)

    for trace in box.data:
        fig.add_trace(trace, row=1, col=2)


    violin = px.violin(df, y=column, box=True)

    for trace in violin.data:
        fig.add_trace(trace, row=1, col=3)


    ecdf = px.ecdf(df, x=column)

    for trace in ecdf.data:
        fig.add_trace(trace, row=1, col=4)


    strip = px.strip(df, x=column)

    for trace in strip.data:
        fig.add_trace(trace, row=1, col=5)


    fig.update_layout(
        title_text=f"Numerical Univariate Analysis: {column}",
        height=500,
        showlegend=False
    )

    return fig
```

```python
def numerical_univariate_plot(df, column, plot_type="histogram"):

    if plot_type == "histogram":

        fig = px.histogram(df, x=column, title=f"Distribution of {column}")

    elif plot_type == "box":

        fig = px.box(df, y=column, title=f"Box Plot of {column}")

    elif plot_type == "violin":

        fig = px.violin(df, y=column, box=True, title=f"Violin Plot of {column}")

    elif plot_type == "ecdf":

        fig = px.ecdf(df, x=column, title=f"ECDF of {column}")

    elif plot_type == "strip":

        fig = px.strip(df, x=column, title=f"Strip Plot of {column}")

    return fig
```

```python
def numerical_bivariate_all_plots(df, column1, column2):

    fig = make_subplots(
        rows=1,
        cols=3,
        subplot_titles=["Scatter Plot", "Density Heatmap", "Density Contour"],
        specs=[[{"type": "xy"}, {"type": "xy"}, {"type": "xy"}]]
    )

    scatter = px.scatter(df, x=column1, y=column2)

    for trace in scatter.data:
        fig.add_trace(trace, row=1, col=1)
    fig.update_xaxes(title_text=column1, row=1, col=1)
    fig.update_yaxes(title_text=column2, row=1, col=1)

    density_heatmap = px.density_heatmap(df, x=column1, y=column2)

    for trace in density_heatmap.data:
        fig.add_trace(trace, row=1, col=2)
    fig.update_xaxes(title_text=column1, row=1, col=2)

    fig.update_yaxes(title_text=column2, row=1, col=2)

    density_contour = px.density_contour(df, x=column1, y=column2)

    for trace in density_contour.data:
        fig.add_trace(trace, row=1, col=3)
    fig.update_xaxes(title_text=column1, row=1, col=3)
    fig.update_yaxes(title_text=column2, row=1, col=3)

    fig.update_layout(
        title_text=f"Numerical Bivariate Analysis: {column1} vs {column2}",
        height=500,
        showlegend=False
    )

    return fig
```

```python
def numerical_bivariate_plot(df, column1, column2, plot_type="scatter"):

    if plot_type == "scatter":

        fig = px.scatter(df, x=column1, y=column2, title=f"{column1} vs {column2}")

    elif plot_type == "density_heatmap":

        fig = px.density_heatmap(df, x=column1, y=column2, title=f"Density Heatmap: {column1} vs {column2}")

    elif plot_type == "density_contour":

        fig = px.density_contour(df, x=column1, y=column2, title=f"Density Contour: {column1} vs {column2}")

    return fig
```

این تابع با متد Pearson بین ستون های عددی correlation را محاسبه می کند.

```python
def numerical_correlation_heatmap(df):
    
    corr = df.corr()

    fig = px.imshow(corr, text_auto=".2f", color_continuous_scale="RdBu_r", zmin=-1, zmax=1, title="Correlation Heatmap")

    return fig
```
